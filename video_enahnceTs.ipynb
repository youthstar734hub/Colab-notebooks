```python
# BasicSR clone
!git clone --depth 1 https://github.com/XPixelGroup/BasicSR.git
%cd BasicSR

# setup.py fix (Python 3.13)
code = '''
from pathlib import Path
p = Path("setup.py")
text = p.read_text()

old = """def get_version():
    with open(version_file, 'r') as f:
        exec(compile(f.read(), version_file, 'exec'))
    return locals()['__version__']"""

new = """def get_version():
    with open(version_file, 'r') as f:
        d = {}
        exec(compile(f.read(), version_file, 'exec'), d)
    return d['__version__']"""

if old in text:
    p.write_text(text.replace(old, new))
    print("✅ BasicSR setup.py fixed successfully")
else:
    print("Already fixed or different version found")
'''

with open("fix_setup.py", "w") as f:
    f.write(code)

!python fix_setup.py

# Install
!pip install -q -r requirements.txt
!python setup.py develop

%cd /content
print("✅ BasicSR installed successfully!")
```
